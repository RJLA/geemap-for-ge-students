# What is Google Earth Engine

Google Earth Engine (GEE) is a cloud platform with a large catalog of satellite imagery and geospatial data. Since its public launch in 2010, it has become widely used for environmental studies at all scales.

Research using GEE has grown rapidly. As of 2022, over 1,000 journal articles mention it in the title, and nearly 3,000 mention it in the title or abstract—a massive increase from just 4 publications in 2014.

Google Earth Engine (GEE) offers JavaScript and Python APIs. The JavaScript API has strong support and a built-in editor, but the Python API lacks interactive visualization tools and documentation.

# What is GEEMAP
geemap was created to bridge this gap. Built on open-source Python libraries, it lets users analyze and visualize GEE data in Jupyter notebooks with minimal code. It’s ideal for researchers, students, and anyone moving from the JavaScript API to Python. geemap also includes a GUI for converting JavaScript code to Python, saving time and effort.

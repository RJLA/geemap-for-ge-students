# Earth Engine data reductions

Reducers can be used to aggregate data over space, time, bands, lists, and other data structures in Earth Engine. For example, given a list of numbers, you might want to calculate some
descriptive statistics on the list, such as the minimum, maximum, sum, mean, median, and
standard deviation. These operations are what reducers do, taking an input dataset and produce a single output.

```python
# List reductions
values = ee.List.sequence(1, 10)
print(values.getInfo())
```

```python
# calculate the number of values in the list, use the ee.Reducer.count() function:
count = values.reduce(ee.Reducer.count())
print(count.getInfo()) # 10
```

```python
# To find out the minimum value in the list, use the ee.Reducer.min() function
min_value = values.reduce(ee.Reducer.min())
print(min_value.getInfo()) # 1
```

```python
# To find out the maximum value in the list, use the ee.Reducer.max() function:
max_value = values.reduce(ee.Reducer.max())
print(max_value.getInfo()) # 10
```

```python
# To find out the minimum and maximum values in the list in one round, use ee.Reducer.minMax():
min_max_value = values.reduce(ee.Reducer.minMax())
print(min_max_value.getInfo())
```

```python
# To calculate the mean of the list, use the ee.Reducer.mean() function:
mean_value = values.reduce(ee.Reducer.mean())
print(mean_value.getInfo()) # 5.5
```

```python
# To calculate the median of the list, use the ee.Reducer.median() function:
median_value = values.reduce(ee.Reducer.median())
print(median_value.getInfo()) # 5.5
```

```python
# To calculate the sum of the list, use the ee.Reducer.sum() function:
sum_value = values.reduce(ee.Reducer.sum())
print(sum_value.getInfo()) # 55
```

```python
# To calculate the standard deviation of the list, use the ee.Reducer.stdDev() function:
std_value = values.reduce(ee.Reducer.stdDev())
print(std_value.getInfo()) # 2.8723
```

# ImageCollection reductions

Reducing an ImageCollection is one of the most common operations in Earth Engine. Assume
that you have a series of images as an ImageCollection covering an area of interest, and you
want to produce a good-quality cloud-free image of the area based on the existing images. In
this case, you might want to use the imageCollection.reduce() or imageCollection.median()
methods to reduce the ImageCollection to a single image. The output is computed pixel-wise,
such that each pixel in the output is composed of the median value of all the images in the
collection at that location

```python
# Load an image collection, filtered so it's not too much data.
Map = geemap.Map()
START_DATE = "2023-01-01"
END_DATE = "2024-12-31"
X = 121.07
Y = 14.66
Z = 16
S2_IMAGE_COLLECTION = "COPERNICUS/S2_HARMONIZED"

collection = (
ee.ImageCollection(S2_IMAGE_COLLECTION)
.filterDate(START_DATE, END_DATE)
.filter(ee.Filter.lt("CLOUDY_PIXEL_PERCENTAGE", 5))
)
#Compute the median in each band, each pixel.
# Band names are B1_median, B2_median, etc.
median = collection.reduce(ee.Reducer.median())

vis = {
"min": 0.0,
"max": 3000,
"bands": ['B4_median', 'B3_median', 'B2_median'],
}
Map.setCenter(X, Y, Z)
Map.addLayer(median, vis, "Sentinel-2")
Map
```

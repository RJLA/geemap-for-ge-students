```python
# import needed libraries
import ee
import geemap
```

```python
# Initialize the Earth Engine Python API
geemap.ee_initialize(project='geemap-for-ge-students')
```

```python
# PLot using plotting tool
# Visualize Fale color using UI
Map = geemap.Map()
S2_IMAGE_COLLECTION = "COPERNICUS/S2_HARMONIZED"
X = 121.07
Y = 14.66
Z = 16
START_DATE = "2023-01-01"
END_DATE = "2024-12-31"

collection = (
ee.ImageCollection(S2_IMAGE_COLLECTION)
.filterDate(START_DATE, END_DATE)
.filter(ee.Filter.lt("CLOUDY_PIXEL_PERCENTAGE", 5))
)
image = collection.median().select(["B2","B3","B4","B8"])
vis = {
"min": 0.0,
"max": 3000,
"bands": ['B4', 'B3', 'B2'],
}
Map.setCenter(X, Y, Z)
Map.addLayer(image, vis, "Sentinel-2")
Map
```

```python
Map = geemap.Map()
Map.add_basemap('HYBRID')
lc = ee.ImageCollection('ESA/WorldCover/v200').first();
landcover = lc.select('Map')
Map.addLayer(landcover, {}, 'LC')
Map.setCenter(X, Y, Z)
Map.add_legend(
title="ESA_WorldCover", builtin_legend='ESA_WorldCover', height='200px', position='bottomright'
)
Map
```

```python
Map = geemap.Map()
dem = ee.Image('USGS/SRTMGL1_003')
vis_params = {
'min': 0,
'max': 4000,
'palette': ['006633', 'E5FFCC', '662A00', 'D8D8D8', 'F5F5F5'],
}
Map.addLayer(dem, vis_params, 'SRTM DEM')
Map.add_colorbar(vis_params, label="Elevation (m)", layer_name="SRTM DEM")
Map
```

```python
# Split-panel maps
Map = geemap.Map(zoom=4, height=600)
S2_IMAGE_COLLECTION = "COPERNICUS/S2_HARMONIZED"
X = 120.9977
Y = 14.0113
Z = 16
START_DATE_1 = "2019-01-01"
END_DATE_1 = "2019-12-31"

START_DATE_2 = "2020-01-01"
END_DATE_2 = "2020-12-31"

# Make an image of the Taal before eruption
collection_1 = (
ee.ImageCollection(S2_IMAGE_COLLECTION)
.filterDate(START_DATE_1, END_DATE_1)
.filter(ee.Filter.lt("CLOUDY_PIXEL_PERCENTAGE", 5))
)
taal_before = collection_1.median().select(["B2","B3","B4"])

# Make an image of the Taal before eruption
collection_2 = (
ee.ImageCollection(S2_IMAGE_COLLECTION)
.filterDate(START_DATE_2, END_DATE_2)
.filter(ee.Filter.lt("CLOUDY_PIXEL_PERCENTAGE", 5))
)
taal_after = collection_2.median().select(["B2","B3","B4"])

Map.setCenter(X, Y, Z)

vis = {
"min": 0.0,
"max": 3000,
"bands": ['B4', 'B3', 'B2'],
}

taal_before_layer = geemap.ee_tile_layer(taal_before, vis, 'Taal Before Eruption')
taal_after_layer = geemap.ee_tile_layer(taal_after, vis, 'Taal After Eruption')

Map.split_map(taal_before_layer, taal_after_layer, add_close_button=True)
Map
```

```python
# Split-panel maps
Map = geemap.Map(zoom=4, height=600)
S2_IMAGE_COLLECTION = "COPERNICUS/S2_HARMONIZED"
X = 120.9977
Y = 14.0113
Z = 16
START_DATE = "2019-01-01"
END_DATE = "2019-12-31"

# Make an image of the Taal before eruption
collection = (
ee.ImageCollection(S2_IMAGE_COLLECTION)
.filterDate(START_DATE, END_DATE)
.filter(ee.Filter.lt("CLOUDY_PIXEL_PERCENTAGE", 5))
)
image = collection.median()

vis_params = [
{'bands': ['B4', 'B3', 'B2'], 'min': 0, 'max': 3000, 'gamma': 1.3},
{'bands': ['B8', 'B11', 'B4'], 'min': 0, 'max': 3000, 'gamma': 1.3},
{'bands': ['B8', 'B4', 'B3'], 'min': 0, 'max': 3000, 'gamma': 1.3},
{'bands': ['B12', 'B12', 'B4'], 'min': 0, 'max': 3000, 'gamma': 1.3},
]

labels = [
'Natural Color (B4/B3/B2)',
'Land/Water (B8/B11/B4)',
'Color Infrared (B8/B4/B3)',
'Vegetation (B12/B11/B4)',
]
geemap.linked_maps(
rows=2,
cols=2,
height="300px",
center=[38.4151, 21.2712],
zoom=12,
ee_objects=[image],
vis_params=vis_params,
labels=labels,
label_position="topright",
)
```
